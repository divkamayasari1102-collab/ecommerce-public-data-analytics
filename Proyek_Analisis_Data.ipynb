## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

### Cleaning Data
Pada tahap pembersihan data (Cleaning Data) ini, saya mengeksekusi rencana aksi yang telah dirumuskan pada tahap penilaian (Assessing Data). Langkah ini sangat penting untuk memastikan data yang akan dieksplorasi sudah terbebas dari anomali tipe data serta ketidaklengkapan nilai, sehingga hasil analisis dan visualisasi nantinya bersifat valid dan dapat dipertanggungjawabkan.

#### Fixing Problems in Datasets
Pada sub-tahap ini, saya berfokus untuk menyelesaikan dua masalah kualitas data utama yang ditemukan:
1. **Mengubah tipe data string menjadi datetime** pada lima kolom penanda waktu tabel `orders_df` menggunakan fungsi `pd.to_datetime()`. Hal ini diperlukan agar operasi berbasis tanggal dan waktu dapat dilakukan dengan lancar.
2. **Mengatasi missing values** pada kolom `product_category_name` di tabel `products_df` dengan menerapkan metode imputasi menggunakan nilai default `"unknown"` lewat fungsi `.fillna()`. Dengan begitu, data transaksi produk terkait tetap utuh saat dilakukan penggabungan tabel.

```python
# 1. Memperbaiki tipe data string/object menjadi datetime pada tabel orders_df
datetime_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for column in datetime_columns:
    orders_df[column] = pd.to_datetime(orders_df[column])

# 2. Menangani missing values pada nama kategori produk di tabel products_df
# Melakukan imputasi dengan nilai default "unknown" agar tidak terbuang saat merge
products_df["product_category_name"] = products_df["product_category_name"].fillna("unknown")

# 3. Memastikan perubahan dengan memeriksa tipe data dan sisa missing values
print("--- Tipe Data Kolom Tanggal Setelah Diperbaiki ---")
print(orders_df[datetime_columns].dtypes)

print("\n--- Sisa Missing Values Kategori Produk ---")
print(f"Missing values pada product_category_name: {products_df['product_category_name'].isna().sum()}")
```

#### Insight / Rangkuman Proses Cleaning Data:
1. **Konversi Tipe Data Sukses**: Seluruh kolom penanda waktu pada tabel `orders_df` telah berhasil diubah menjadi tipe data `datetime64[ns]`. Langkah ini krusial agar saya dapat mengekstrak tren bulanan dan menghitung durasi pengiriman barang secara akurat tanpa error.
2. **Penanganan Missing Values Berhasil**: Masalah nilai yang hilang (*missing values*) pada kolom `product_category_name` telah diatasi dengan metode imputasi menggunakan nilai bawaan (*default*) yaitu `"unknown"`. Hal ini memastikan tidak ada data transaksi produk yang terbuang secara tidak sengaja saat proses penggabungan tabel (*merge*) nanti.
3. **Integritas Data Terjaga**: Dataset kini berada dalam kondisi optimal, konsisten, dan sepenuhnya siap untuk melangkah ke tahap Eksplorasi Data (EDA).

## Exploratory Data Analysis (EDA)
Proses kritis dalam analisis data yang bertujuan untuk memahami karakteristik dataset, menemukan pola tersembunyi, mendeteksi anomali, serta menguji hipotesis awal melalui ringkasan statistik dan agregasi data. Pada tahap ini, saya akan melakukan eksplorasi data secara terarah untuk menjawab kelima pertanyaan bisnis yang telah ditetapkan, dengan berfokus pada rentang waktu operasional tahun 2017 hingga 2018.

#### Explore and Aggregate Dataset for Business Questions
Untuk mempermudah proses analisis, langkah pertama yang dilakukan adalah menggabungkan (merge) seluruh tabel data yang relevan ke dalam satu DataFrame utama bernama `main_df`. Setelah data terkonsolidasi, saya akan membuat kolom baru untuk ekstraksi bulan pembelian (`order_month`) dan durasi pengiriman (`delivery_time_days`). Selanjutnya, saya melakukan proses agregasi kelompok data untuk mengekstrak informasi tren bulanan, performa kategori produk, skor ulasan, serta preferensi metode pembayaran.

```python
# 1. Menambahkan kolom tahun-bulan untuk memudahkan analisis tren berbasis waktu
orders_df['order_month'] = orders_df['order_purchase_timestamp'].dt.to_period('M')

# 2. Menghitung rata-rata waktu pengiriman dalam satuan hari (Waktu sampai ke pelanggan dikurangi waktu pembelian)
orders_df['delivery_time_days'] = (orders_df['order_delivered_customer_date'] - orders_df['order_purchase_timestamp']).dt.days

# 3. Menggabungkan seluruh tabel menjadi satu DataFrame utama (main_df)
main_df = orders_df.merge(order_items_df, on="order_id", how="left")
main_df = main_df.merge(order_payments_df, on="order_id", how="left")
main_df = main_df.merge(order_reviews_df, on="order_id", how="left")
main_df = main_df.merge(products_df, on="product_id", how="left")
main_df = main_df.merge(product_category_name_translation_df, on="product_category_name", how="left")

# Menggunakan nama kategori bahasa Inggris jika tersedia, jika tidak tetap gunakan nama asli/unknown
main_df['product_category'] = main_df['product_category_name_english'].fillna(main_df['product_category_name'])

# Filter data secara ketat hanya untuk rentang waktu tahun 2017 hingga 2018 sesuai pertanyaan bisnis
main_df = main_df[(main_df['order_purchase_timestamp'].dt.year >= 2017) & (main_df['order_purchase_timestamp'].dt.year <= 2018)]

print(f"--- Hasil Eksplorasi Gabungan Data (main_df) Periode 2017-2018 ---")
print(f"Total baris data gabungan: {main_df.shape[0]}")
print(f"Total kolom data gabungan: {main_df.shape[1]}\n")

# === PROSES AGREGASI EDA UNTUK MENJAWAB 5 PERTANYAAN BISNIS ===

print("1. AGREGASI TREN BULANAN (Pertanyaan 1):")
monthly_trend = main_df.groupby('order_month').agg(
    total_orders=('order_id', 'nunique'),
    total_revenue=('price', 'sum')
).reset_index()
print(monthly_trend.head(3))

print("\n2. AGREGASI KATEGORI PRODUK TOP & BOTTOM (Pertanyaan 2):")
product_performance = main_df.groupby('product_category').agg(
    volume_penjualan=('order_item_id', 'count')
).sort_values(by='volume_penjualan', ascending=False).reset_index()
print("- Top 3 Kategori:")
print(product_performance.head(3))
print("- Bottom 3 Kategori:")
print(product_performance.tail(3))

print("\n3. AGREGASI SKOR REVIW / KEPUASAN (Pertanyaan 3):")
review_dist = main_df['review_score'].value_counts().sort_index()
print(review_dist)

print("\n4. AGREGASI DURASI PENGIRIMAN BULANAN (Pertanyaan 4):")
monthly_delivery = main_df.groupby(main_df['order_purchase_timestamp'].dt.to_period('M'))['delivery_time_days'].mean().reset_index()
print(monthly_delivery.head(3))

print("\n5. AGREGASI METODE PEMBAYARAN (Pertanyaan 5):")
payment_usage = main_df['payment_type'].value_counts()
print(payment_usage)
```

#### Insight / Rangkuman Proses Exploratory Data Analysis (EDA):
1. **Konsolidasi Data Sukses**: Seluruh informasi dari tabel pesanan, item, pembayaran, ulasan, hingga terjemahan nama kategori produk telah menyatu ke dalam objek `main_df` yang dibatasi secara ketat pada periode transaksi tahun 2017-2018.
2. **Ketersediaan Data Agregasi**: Struktur data hasil agregasi kelompok (seperti tren bulanan, performa volume produk, sebaran kepuasan pelanggan, rata-rata durasi pengiriman, dan dominasi jenis pembayaran) telah berhasil diekstrak dan diverifikasi nilainya.
3. **Kesiapan Visualisasi**: Data agregat ini telah siap sepenuhnya untuk dikirim ke tahap berikutnya, yaitu pembuatan grafik visualisasi yang sinkron pada bab *Visualization & Explanatory Analysis*.

## Visualization & Explanatory Analysis
Pada tahap *Visualization & Explanatory Analysis* ini, saya akan mengonversi hasil agregasi data dari tahap EDA ke dalam bentuk visualisasi grafik yang interaktif, bersih, dan informatif. Pemilihan jenis grafik disesuaikan dengan karakteristik data untuk mempermudah penarikan kesimpulan. #### **Pertanyaan 1:** Bagaimana tren total penjualan dan pendapatan (revenue) operasional e-commerce secara bulanan selama periode tahun 2017 hingga 2018?
Pada bagian ini, saya menyajikan data tren performa penjualan (jumlah pesanan unik) serta total pendapatan operasional platform e-commerce yang dihitung secara bulanan.

```python
# Menyiapkan data tren bulanan dari hasil EDA sebelumnya
monthly_data = main_df.groupby('order_month').agg(
    total_orders=('order_id', 'nunique'),
    total_revenue=('price', 'sum')
).reset_index()

# Mengonversi format index order_month menjadi string agar rapi saat ditampilkan di sumbu X
monthly_data['order_month_str'] = monthly_data['order_month'].astype(str)

# Membuat kanvas grafik ganda (Subplots: Kiri untuk Order, Kanan untuk Revenue)
fig, ax = plt.subplots(nrows=1, ncols=2, figsize=(24, 6))

# Grafik 1: Tren Total Penjualan (Orders)
sns.lineplot(
    x="order_month_str",
    y="total_orders",
    data=monthly_data,
    marker="o",
    linewidth=3,
    color="#2A9D8F",
    ax=ax[0]
)
ax[0].set_title("Tren Total Penjualan Bulanan (2017-2018)", loc="center", fontsize=18, fontweight="bold", pad=15)
ax[0].set_xlabel("Bulan Pembelian", fontsize=14, labelpad=10)
ax[0].set_ylabel("Jumlah Pesanan Terkirim", fontsize=14)
ax[0].tick_params(axis='x', rotation=45, labelsize=12)
ax[0].tick_params(axis='y', labelsize=12)

# Grafik 2: Tren Total Pendapatan (Revenue)
sns.lineplot(
    x="order_month_str",
    y="total_revenue",
    data=monthly_data,
    marker="s",
    linewidth=3,
    color="#E76F51",
    ax=ax[1]
)
ax[1].set_title("Tren Total Pendapatan Bulanan (2017-2018)", loc="center", fontsize=18, fontweight="bold", pad=15)
ax[1].set_xlabel("Bulan Pembelian", fontsize=14, labelpad=10)
ax[1].set_ylabel("Total Pendapatan (Rupiah/Mata Uang)", fontsize=14)
ax[1].tick_params(axis='x', rotation=45, labelsize=12)
ax[1].tick_params(axis='y', labelsize=12)

plt.tight_layout()
plt.show()
```

#### **Pertanyaan 2:** Kategori Produk Apa Saja yang Mencatatkan Volume Penjualan Tertinggi dan Terendah (Top 5 & Bottom 5) Selama Periode Tahun 2017 Hingga 2018?
Pada bagian ini, saya menyajikan visualisasi komparatif untuk mengidentifikasi 5 kategori produk yang paling diminati (Top 5) dan 5 kategori produk dengan volume penjualan paling sedikit (Bottom 5).


```python
# Menyiapkan data agregasi produk dari main_df
product_perf = main_df.groupby('product_category').agg(
    volume_penjualan=('order_item_id', 'count')
).reset_index()

# Memisahkan Top 5 dan Bottom 5 kategori produk
top_5_products = product_perf.sort_values(by='volume_penjualan', ascending=False).head(5)
bottom_5_products = product_perf.sort_values(by='volume_penjualan', ascending=True).head(5)

# Membuat kanvas grafik berdampingan (1 baris, 2 kolom)
fig, ax = plt.subplots(nrows=1, ncols=2, figsize=(24, 8))

# Grafik Kiri: Top 5 Kategori Produk
sns.barplot(
    x="volume_penjualan",
    y="product_category",
    data=top_5_products,
    palette="viridis",
    ax=ax[0]
)
ax[0].set_title("Top 5 Kategori Produk Tertinggi (2017-2018)", loc="center", fontsize=18, fontweight="bold", pad=15)
ax[0].set_xlabel("Volume Penjualan (Jumlah Item)", fontsize=14, labelpad=10)
ax[0].set_ylabel("Kategori Produk", fontsize=14)
ax[0].tick_params(axis='both', labelsize=12)

# Grafik Kanan: Bottom 5 Kategori Produk
sns.barplot(
    x="volume_penjualan",
    y="product_category",
    data=bottom_5_products,
    palette="rocket",
    ax=ax[1]
)
ax[1].set_title("Bottom 5 Kategori Produk Terendah (2017-2018)", loc="center", fontsize=18, fontweight="bold", pad=15)
ax[1].set_xlabel("Volume Penjualan (Jumlah Item)", fontsize=14, labelpad=10)
ax[1].set_ylabel(None) # Dikosongkan agar tidak menumpuk di tengah
ax[1].tick_params(axis='both', labelsize=12)

plt.tight_layout()
plt.show()
```

#### **Pertanyaan 3:** Bagaimana distribusi tingkat kepuasan pelanggan berdasarkan skor ulasan (review score) pada seluruh transaksi yang selesai selama periode tahun 2017 hingga 2018?
Pada bagian ini, saya menyajikan data sebaran skor ulasan dari skala 1 sampai 5 untuk mengevaluasi tingkat kepuasan pelanggan secara keseluruhan.

```python
# Menyiapkan data distribusi skor ulasan dari main_df
review_counts = main_df['review_score'].dropna().value_counts().sort_index().reset_index()
review_counts.columns = ['review_score', 'count']

# Menentukan warna: warna mencolok untuk nilai tertinggi (skor 5) agar grafik lebih informatif
colors = ["#E76F51" if score != 5 else "#2A9D8F" for score in review_counts['review_score']]

# Membuat visualisasi grafik batang
plt.figure(figsize=(10, 6))
sns.barplot(
    x="review_score",
    y="count",
    data=review_counts,
    palette=colors
)

# Menambahkan label angka di atas setiap bar grafik
for index, row in review_counts.iterrows():
    plt.text(
        x=index,
        y=row['count'] + (row['count'] * 0.01),
        s=f"{int(row['count']):,}",
        ha="center",
        va="bottom",
        fontsize=11,
        fontweight="bold"
    )

plt.title("Distribusi Skor Ulasan/Kepuasan Pelanggan (2017-2018)", fontsize=16, fontweight="bold", pad=15)
plt.xlabel("Skor Ulasan (Review Score)", fontsize=12, labelpad=10)
plt.ylabel("Jumlah Transaksi", fontsize=12)
plt.tick_params(axis='both', labelsize=11)

plt.tight_layout()
plt.show()
```

#### **Pertanyaan 4:** Berapa rata-rata waktu pengiriman barang (delivery time) dari penjual ke pembeli secara bulanan, dan apakah performanya membaik di tahun 2018?
Pada bagian ini, saya menyajikan grafik tren durasi pengiriman barang dari pintu penjual hingga sampai ke tangan pembeli (dalam satuan hari) untuk mengukur efisiensi logistik.


```python
# Menyiapkan data durasi pengiriman rata-rata bulanan dari main_df
delivery_data = main_df.groupby('order_month').agg(
    avg_delivery_days=('delivery_time_days', 'mean')
).reset_index()

# Mengonversi format index order_month menjadi string agar rapi di sumbu X
delivery_data['order_month_str'] = delivery_data['order_month'].astype(str)

# Membuat visualisasi grafik tren garis
plt.figure(figsize=(12, 6))
sns.lineplot(
    x="order_month_str",
    y="avg_delivery_days",
    data=delivery_data,
    marker="o",
    linewidth=3,
    color="#457B9D"
)

# Menambahkan garis bantu rata-rata keseluruhan untuk mempermudah analisis perbandingan tahunan
overall_mean = delivery_data['avg_delivery_days'].mean()
plt.axhline(overall_mean, color='#E63946', linestyle='--', linewidth=1.5, label=f'Rata-rata: {overall_mean:.1f} Hari')

plt.title("Tren Rata-rata Durasi Pengiriman Barang Bulanan (2017-2018)", fontsize=16, fontweight="bold", pad=15)
plt.xlabel("Bulan Pembelian", fontsize=12, labelpad=10)
plt.ylabel("Rata-rata Durasi (Hari)", fontsize=12)
plt.xticks(rotation=45)
plt.tick_params(axis='both', labelsize=11)
plt.legend(fontsize=11)

plt.tight_layout()
plt.show()
```

## Analisis Lanjutan: RFM Analysis (Recency, Frequency, Monetary)
Analisis RFM adalah teknik segmentasi pelanggan yang digunakan untuk mengevaluasi perilaku transaksi berdasarkan tiga metrik kuantitatif:
* **Recency (R)**: Mengukur seberapa baru pelanggan melakukan transaksi terakhir (dalam satuan hari). Semakin kecil nilainya, semakin baik.
* **Frequency (F)**: Mengukur seberapa sering pelanggan melakukan pembelian dalam periode tertentu. Semakin besar nilainya, semakin aktif.
* **Monetary (M)**: Mengukur total nilai finansial atau uang yang dihabiskan pelanggan untuk berbelanja. Semakin besar nilainya, semakin bernilai pelanggan tersebut.

Melalui analisis ini, saya akan mengidentifikasi 5 pelanggan terbaik (*Top 5 Customers*) untuk masing-masing dimensi RFM pada periode 2017-2018.


```python
# 1. Menentukan tanggal referensi (hari terakhir transaksi di dataset + 1 hari)
recent_date = main_df['order_purchase_timestamp'].max() + pd.Timedelta(days=1)

# 2. Menghitung nilai RFM berdasarkan customer_id
rfm_df = main_df.groupby('customer_id').agg(
    recency=('order_purchase_timestamp', lambda x: (recent_date - x.max()).days),
    frequency=('order_id', 'nunique'),
    monetary=('price', 'sum')
).reset_index()

# 3. Mengambil Top 5 Pelanggan untuk masing-masing dimensi RFM
top_5_recency = rfm_df.sort_values(by="recency", ascending=True).head(5)
top_5_frequency = rfm_df.sort_values(by="frequency", ascending=False).head(5)
top_5_monetary = rfm_df.sort_values(by="monetary", ascending=False).head(5)

# Menyederhanakan string Customer ID agar grafik tidak menumpuk dan mudah dibaca (diambil 8 karakter awal)
top_5_recency['customer_short'] = top_5_recency['customer_id'].str[:8]
top_5_frequency['customer_short'] = top_5_frequency['customer_id'].str[:8]
top_5_monetary['customer_short'] = top_5_monetary['customer_id'].str[:8]

# 4. Membuat Visualisasi Grafik Batang Bersandingan (1 Baris, 3 Kolom)
fig, ax = plt.subplots(nrows=1, ncols=3, figsize=(24, 8))

# Grafik 1: Top 5 Customers by Recency
sns.barplot(
    y="recency",
    x="customer_short",
    data=top_5_recency,
    palette="BuGn_r",
    ax=ax[0]
)
ax[0].set_title("Top 5 Customers by Recency (Days)", fontsize=16, fontweight="bold", pad=15)
ax[0].set_xlabel("Customer ID (Short)", fontsize=12)
ax[0].set_ylabel("Recency (Hari)", fontsize=12)
ax[0].tick_params(axis='x', rotation=30, labelsize=11)

# Grafik 2: Top 5 Customers by Frequency
sns.barplot(
    y="frequency",
    x="customer_short",
    data=top_5_frequency,
    palette="Oranges_r",
    ax=ax[1]
)
ax[1].set_title("Top 5 Customers by Frequency", fontsize=16, fontweight="bold", pad=15)
ax[1].set_xlabel("Customer ID (Short)", fontsize=12)
ax[1].set_ylabel("Frequency (Jumlah Order)", fontsize=12)
ax[1].tick_params(axis='x', rotation=30, labelsize=11)

# Grafik 3: Top 5 Customers by Monetary
sns.barplot(
    y="monetary",
    x="customer_short",
    data=top_5_monetary,
    palette="Blues_r",
    ax=ax[2]
)
ax[2].set_title("Top 5 Customers by Monetary (Revenue)", fontsize=16, fontweight="bold", pad=15)
ax[2].set_xlabel("Customer ID (Short)", fontsize=12)
ax[2].set_ylabel("Monetary (Total Belanja)", fontsize=12)
ax[2].tick_params(axis='x', rotation=30, labelsize=11)

plt.suptitle("Analisis Lanjutan: Identifikasi 5 Pelanggan Terbaik Berdasarkan Parameter RFM", fontsize=20, fontweight="bold", y=1.02)
plt.tight_layout()
plt.show()
```

#### **Insight dari Hasil Analisis Lanjutan RFM:**
Berdasarkan hasil visualisasi grafik segmentasi pelanggan menggunakan parameter RFM (Recency, Frequency, Monetary) di atas, saya dapat mengambil beberapa poin insight penting:

1. **Parameter Recency (Kekinian Transaksi)**: Grafik berhasil mengisolasi 5 pelanggan teratas yang memiliki nilai *recency* sangat kecil (0-1 hari). Hal ini menandakan bahwa kelompok pengguna tersebut baru saja melakukan transaksi aktif di platform e-commerce. Pelanggan pada segmen ini sangat potensial untuk diberikan penawaran promosi produk pelengkap (*cross-selling*) karena responsivitas mereka yang masih tinggi.
2. **Parameter Frequency (Kerapatan Transaksi)**: Mayoritas basis pelanggan pada e-commerce ini cenderung melakukan transaksi tunggal (1 kali). Namun, melalui grafik ini, saya berhasil mengidentifikasi 5 pengguna loyal yang melakukan transaksi berulang secara intensif melebihi rata-rata pengguna umum dalam periode tahun 2017-2018.
3. **Parameter Monetary (Nilai Finansial)**: Grafik berhasil memetakan segmen pelanggan *"Whales"*, yaitu 5 pengguna dengan kontribusi pengeluaran atau total nilai belanja terbesar di platform. Kelompok ini adalah penggerak utama *revenue* operasional e-commerce yang wajib dijaga retensinya melalui program loyalitas eksklusif (*loyalty program*).
